# Statistical tests

```python
from scipy.stats import wilcoxon 

models = [
    {'name': 'CSP',
        'binary': [0.54, 0.54, 0.51, 0.57, 0.58, 0.49, 0.52, 0.46, 0.51],
        '4-class': [0.27, 0.26, 0.29, 0.28, 0.28, 0.27, 0.24, 0.29, 0.26] 
    }, 
    {'name': 'EEGNet',
        'binary': [0.83, 0.59, 0.83, 0.69, 0.76, 0.58, 0.84, 0.74, 0.60],
        '4-class': [0.56, 0.29, 0.51, 0.39, 0.36, 0.34, 0.45, 0.43, 0.40]  
    },
    {'name': 'MiniRocket',
        'binary': [0.61, 0.62, 0.53, 0.54, 0.54, 0.57, 0.62, 0.68, 0.55] ,
        '4-class': [0.47, 0.42, 0.41, 0.30, 0.33, 0.34, 0.40, 0.35, 0.39]   
    } 
] 
```

```python
print("===== BINARY =====")
for i in range(3):
    if i == 2:
        sec = 0
    else:
        sec = i+1
    print(f"{models[i]['name']} vs {models[sec]['name']}")
    stat, p = wilcoxon(models[i]['binary'], models[sec]['binary']) 
    print(f"p-value: {p:.4f}")

print("===== 4 CLASS =====")
for i in range(3):
    if i == 2:
        sec = 0
    else:
        sec = i+1
    print(f"{models[i]['name']} vs {models[sec]['name']}")
    stat, p = wilcoxon(models[i]['4-class'], models[sec]['4-class']) 
    print(f"p-value: {p:.4f}")
```
